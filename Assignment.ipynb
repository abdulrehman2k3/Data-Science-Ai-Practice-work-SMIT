## Read student marks from marks.txt and generate results.
### Format:

    name, marks
### Requirements:
    Calculate average marks
    Find highest and lowest marks
    Handle missing file error
    Skip invalid data rows
    Handle ValueError safely

```python
try:
    file = open("marks.txt", "r")
    lines = file.readlines()
    file.close()
    students = []
    skipped  = 0
    for line in lines:
        line = line.strip()
        if not line:
            continue
        parts = line.split(",")
        if len(parts) != 2:
            print(f"Invalid row skipped: {line}")
            skipped += 1
            continue
        name  = parts[0].strip()
        marks = parts[1].strip()

        try:
            marks = float(marks)
        except ValueError:
            print(f"Invalid marks skipped: {line}")
            skipped += 1
            continue

        students.append({"name": name, "marks": marks})

    if not students:
        print("No valid data found!")
    else:
        total   = 0
        highest = students[0]
        lowest  = students[0]

        for s in students:
            total += s["marks"]
            if s["marks"] > highest["marks"]:
                highest = s
            if s["marks"] < lowest["marks"]:
                lowest = s

        average = total / len(students)

        for s in students:
            print(f"{s['name']} : {s['marks']}")

        print(f"\nTotal Students : {len(students)}")
        print(f"Average Marks  : {average:.2f}")
        print(f"Highest        : {highest['marks']}")
        print(f"Lowest         : {lowest['marks']}")
        print(f"Skipped Rows   : {skipped}")

except FileNotFoundError:
    print("Error: marks.txt file nahi mili!")
```

## Build a banking system using accounts.json
### Format:

    account_id, name, balance
### Requirements:
    Support deposit, withdraw, and transfer
    Prevent negative balance
    If program crashes mid-transfer → must rollback changes
    Log every transaction in bank_log.txt
    Handle corrupted account lines safely
    If file missing → initialize system automatically

```python
import json
import os
from datetime import datetime

ACCOUNT_FILE = 'accounts.json'
LOG_FILE = 'bank_log.txt'
def log_transaction(message):
    """Prints to screen and appends to log file."""
    print(message)
    timestamp = datetime.now().strftime("%Y-%m-%d %H:%M:%S")
    with open(LOG_FILE, 'a') as f:
        f.write(f"[{timestamp}] {message}\n")

def load_accounts():
    """Loads accounts, handles missing files, and skips corrupted data."""
    if not os.path.exists(ACCOUNT_FILE):
        return {} 

    try:
        with open(ACCOUNT_FILE, 'r') as f:
            data = json.load(f)

            clean_data = {}
            for acc_id, info in data.items():
                if isinstance(info, dict) and 'name' in info and 'balance' in info:
                    clean_data[acc_id] = info
            return clean_data
    except Exception:
        log_transaction("WARNING: accounts.json was corrupted. Starting fresh.")
        return {}

def save_accounts(accounts):
    """Saves safely using a temporary file to prevent crash corruption."""
    temp_file = ACCOUNT_FILE + ".tmp"
    with open(temp_file, 'w') as f:
        json.dump(accounts, f, indent=4)
    os.replace(temp_file, ACCOUNT_FILE) 

def deposit(account_id, amount):
    accounts = load_accounts()
    if account_id not in accounts or amount <= 0:
        print("Deposit failed: Invalid account or amount.")
        return
    accounts[account_id]['balance'] += amount
    save_accounts(accounts)
    log_transaction(f"DEPOSIT: {account_id} | +${amount} | New Balance: ${accounts[account_id]['balance']}")
def withdraw(account_id, amount):
    accounts = load_accounts()
    if account_id not in accounts or amount <= 0:
        print("Withdrawal failed: Invalid account or amount.")
        return
    if accounts[account_id]['balance'] < amount:
        log_transaction(f"FAILED WITHDRAWAL: {account_id} tried to withdraw ${amount} (Insufficient Funds)")
        return

    accounts[account_id]['balance'] -= amount
    save_accounts(accounts)
    log_transaction(f"WITHDRAW: {account_id} | -${amount} | New Balance: ${accounts[account_id]['balance']}")

def transfer(from_id, to_id, amount):
    accounts = load_accounts()
    if from_id not in accounts or to_id not in accounts or amount <= 0:
        print("Transfer failed: Invalid accounts or amount.")
        return

    if accounts[from_id]['balance'] < amount:
        log_transaction(f"FAILED TRANSFER: {from_id} -> {to_id} | Insufficient funds.")
        return

    accounts[from_id]['balance'] -= amount
    accounts[to_id]['balance'] += amount
 
    save_accounts(accounts)
    log_transaction(f"TRANSFER: {from_id} sent ${amount} to {to_id}")

def create_account(account_id, name, initial_balance):
    accounts = load_accounts()
    if account_id in accounts:
        print("Account already exists!")
        return
    
    accounts[account_id] = {"name": name, "balance": initial_balance}
    save_accounts(accounts)
    log_transaction(f"CREATED: Account {account_id} for {name} with ${initial_balance}")

if __name__ == "__main__":
    create_account("1", "Alice", 500)
    create_account("2", "Bob", 100)
    deposit("1", 50)
    withdraw("2", 20)
    transfer("1", "2", 100)
```

## Store users in users.json.
### Requirements:
    Add, update, delete users
    Prevent duplicate email or username
    Handle JSON corruption (auto-repair or reset safely)
    Validate email format strictly
    If file missing → create default structure
    Maintain audit log in audit.txt

```python
import json
import re
from datetime import datetime

USERS_FILE = "users.json"
AUDIT_FILE = "audit.txt"

def load_users():
    try:
        with open(USERS_FILE, "r") as file:
            return json.load(file)

    except FileNotFoundError:
        data = {"users": []}
        with open(USERS_FILE, "w") as file:
            json.dump(data, file, indent=4)
        return data

    except json.JSONDecodeError:
        print("JSON corrupted. File reset safely.")
        data = {"users": []}
        with open(USERS_FILE, "w") as file:
            json.dump(data, file, indent=4)
        return data

def save_users(data):
    with open(USERS_FILE, "w") as file:
        json.dump(data, file, indent=4)

def write_log(action):
    with open(AUDIT_FILE, "a") as file:
        file.write(f"{datetime.now()} - {action}\n")

def validate_email(email):
    pattern = r'^[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}$'
    return re.match(pattern, email)

def add_user():
    data = load_users()
    username = input("Enter username: ")
    email = input("Enter email: ")
    if not validate_email(email):
        print("Invalid email format.")
        return
    for user in data["users"]:
        if user["username"] == username:
            print("Username already exists.")
            return
        if user["email"] == email:
            print("Email already exists.")
            return
    data["users"].append({
        "username": username,
        "email": email
    })
    save_users(data)
    write_log(f"User Added: {username}")

    print("User added successfully.")

def update_user():
    data = load_users()

    username = input("Enter username to update: ")

    for user in data["users"]:

        if user["username"] == username:

            new_email = input("Enter new email: ")

            if not validate_email(new_email):
                print("Invalid email format.")
                return

            for u in data["users"]:
                if u["email"] == new_email and u["username"] != username:
                    print("Email already exists.")
                    return

            user["email"] = new_email

            save_users(data)
            write_log(f"User Updated: {username}")

            print("User updated successfully.")
            return

    print("User not found.")

def delete_user():
    data = load_users()

    username = input("Enter username to delete: ")

    for user in data["users"]:

        if user["username"] == username:

            data["users"].remove(user)

            save_users(data)
            write_log(f"User Deleted: {username}")

            print("User deleted successfully.")
            return

    print("User not found.")

def view_users():
    data = load_users()

    print("\nUsers List")

    for user in data["users"]:
        print(
            f"Username: {user['username']}, "
            f"Email: {user['email']}"
        )

while True:

    print("1. Add User")
    print("2. Update User")
    print("3. Delete User")
    print("4. View Users")
    print("5. Exit")

    choice = input("Enter choice: ")

    if choice == "1":
        add_user()

    elif choice == "2":
        update_user()

    elif choice == "3":
        delete_user()

    elif choice == "4":
        view_users()

    elif choice == "5":
        print("Program Ended.")
        break

    else:
        print("Invalid Choice.")
```

## Manage patient data in patients.json
### Format:

    name, age, disease, status
### Requirements:
    Add new patient records using a function
    View all patients safely
    Handle corrupted lines (skip them)
    Raise error if age is negative or invalid
    If file missing, create it automatically

```python
import json
def add_patient():
    try:
        with open("patients.json", "r") as file:
            patients = json.load(file)

    except (FileNotFoundError, json.JSONDecodeError):
        patients = []

    try:
        name = input("Enter patient name: ")
        age = int(input("Enter age: "))

        if age < 0:
            raise ValueError("Age cannot be negative.")

        disease = input("Enter disease: ")
        status = input("Enter status: ")

        patient = {
            "name": name,
            "age": age,
            "disease": disease,
            "status": status
        }

        patients.append(patient)

        with open("patients.json", "w") as file:
            json.dump(patients, file, indent=4)

        print("Patient record added successfully.")

    except ValueError as e:
        print("Error:", e)
def view_patients():

    try:
        with open("patients.json", "r") as file:
            patients = json.load(file)

        print("Patient Records")

        for patient in patients:
            if (
                "name" not in patient
                or "age" not in patient
                or "disease" not in patient
                or "status" not in patient
            ):
                print("Corrupted record skipped.")
                continue

            try:
                age = int(patient["age"])

                if age < 0:
                    raise ValueError

                print(
                    f"Name: {patient['name']}, "
                    f"Age: {patient['age']}, "
                    f"Disease: {patient['disease']}, "
                    f"Status: {patient['status']}"
                )

            except (ValueError, TypeError):
                print("Invalid age value skipped.")

    except FileNotFoundError:
        with open("patients.json", "w") as file:
            json.dump([], file)
        print("patients.json created successfully.")

    except json.JSONDecodeError:
        print("Invalid JSON data.")

add_patient()
view_patients()
```

    Calculate fine:
    1–5 days → 10 per day
    6–10 days → 20 per day
    10 days → 50 per day
### Functions:
    load_records()
    calculate_fines()
    Handle invalid or missing values
    Skip corrupted rows safely
    Handle missing file error

```python
import json
def load_records():
    try:
        with open("library.json", "r") as file:
            records = json.load(file)
        return records
    except FileNotFoundError:
        print("library.json file not found.")
        return []
    except json.JSONDecodeError:
        print("Invalid JSON data.")
def calculate_fines():
    records = load_records()
    if not records:
        return
    print("Library Fine Report")
    for record in records:
        if (
            "student_name" not in record
            or "book_name" not in record
            or "days_late" not in record
        ):
            print("Corrupted record skipped.")
            continue

        try:
            days_late = int(record["days_late"])

            if days_late < 0:
                print("Invalid days_late value skipped.")
                continue
            if 1 <= days_late <= 5:
                fine = days_late * 10

            elif 6 <= days_late <= 10:
                fine = days_late * 20

            elif days_late > 10:
                fine = days_late * 50

            else:
                fine = 0

            print(
                f"Student: {record['student_name']}, "
                f"Book: {record['book_name']}, "
                f"Days Late: {days_late}, "
                f"Fine: {fine}"
            )

        except (ValueError, TypeError):
            print("Invalid record skipped.")

calculate_fines()
```

## Manage appointments in appointments.json.
### Format:

    patient_name, doctor_name, date, status
### Requirements:
#### Functions:
    book_appointment()
    view_appointments()
    Prevent duplicate appointments (same patient + date)
Handle invalid date format safely
Skip corrupted rows
Auto-create file if missing

```python
import json
from datetime import datetime
def book_appointment():
    try:
        with open("appointments.json", "r") as file:
            appointments = json.load(file)
    except (FileNotFoundError, json.JSONDecodeError):
        appointments = []
    patient_name = input("Enter patient name: ")
    doctor_name = input("Enter doctor name: ")
    date = input("Enter appointment date (YYYY-MM-DD): ")
    status = input("Enter status: ")
    try:
        datetime.strptime(date, "%Y-%m-%d")
    except ValueError:
        print("Invalid date format.")
        return
    for appointment in appointments:
        if (
            appointment.get("patient_name") == patient_name
            and appointment.get("date") == date
        ):
            print("Duplicate appointment found.")
            return

    new_appointment = {
        "patient_name": patient_name,
        "doctor_name": doctor_name,
        "date": date,
        "status": status
    }

    appointments.append(new_appointment)

    try:
        with open("appointments.json", "w") as file:
            json.dump(appointments, file, indent=4)

        print("Appointment booked successfully.")

    except IOError:
        print("Error saving appointment.")

def view_appointments():

    try:
        with open("appointments.json", "r") as file:
            appointments = json.load(file)

        print("Appointments List:")

        for appointment in appointments:
            if (
                "patient_name" not in appointment
                or "doctor_name" not in appointment
                or "date" not in appointment
                or "status" not in appointment
            ):
                print("Corrupted record skipped.")
                continue

            print(
                f"Patient: {appointment['patient_name']}, "
                f"Doctor: {appointment['doctor_name']}, "
                f"Date: {appointment['date']}, "
                f"Status: {appointment['status']}"
            )

    except FileNotFoundError:
        print("No appointments found.")

    except json.JSONDecodeError:
        print("Invalid JSON data.")
book_appointment()
view_appointments()
```

## Manage applications in applications.json
### Requirements:
### Format:
    name, skills, experience_years
### Functions:
    submit_application()
    filter_candidates()
    Accept only experience ≥ 1 year
    Handle invalid experience values
    Skip corrupted rows
    Save accepted candidates separately

```python
import json
def submit_application():
    applications = []
    try:
        n = int(input("Enter number of applications: "))
        for i in range(n):
            print(f"Application {i+1}")
            name = input("Enter candidate name: ")
            skills = input("Enter skills (comma separated): ")
            try:
                experience_years = float(input("Enter experience (years): "))
                application = {
                    "name": name,
                    "skills": skills,
                    "experience_years": experience_years
                }
                applications.append(application)
            except ValueError:
                print("Invalid experience value. Application skipped.")
        with open("applications.json", "w") as file:
            json.dump(applications, file, indent=4)
        print("Applications saved successfully.")
    except IOError:
        print("Error writing to file.")
def filter_candidates():
    accepted_candidates = []
    try:
        with open("applications.json", "r") as file:
            data = json.load(file)

        for candidate in data:
            if (
                "name" not in candidate or
                "skills" not in candidate or
                "experience_years" not in candidate
            ):
                print("Corrupted record skipped.")
                continue

            try:
                experience = float(candidate["experience_years"])

                if experience >= 1:
                    accepted_candidates.append(candidate)

            except (ValueError, TypeError):
                print("Invalid experience value skipped.")

        with open("accepted_candidates.json", "w") as file:
            json.dump(accepted_candidates, file, indent=4)

        print("Accepted Candidates:")
        for candidate in accepted_candidates:
            print(
                candidate["name"],
                "-",
                candidate["experience_years"],
                "years"
            )
        print("Accepted candidates saved in accepted_candidates.json")
    except FileNotFoundError:
        print("applications.json not found.")
    except json.JSONDecodeError:
        print("Invalid JSON data.")
    except IOError:
        print("Error reading file.")

submit_application()
filter_candidates()
```
