```python
def detect_outliers(col):
    q1 = np.percentile(col,25)
    q3 = np.percentile(col,75)
    IQR = q3 - q1
    lower_boundary = q1-1.5*IQR
    upper_boundary = q3+1.5*IQR
    return lower_boundary,upper_boundary
```

```python
lb,ub = detect_outliers(data.price)
print(lb,ub)
```

```python
data[(data.price<lb) | (data.price>ub)]
```

```python
data = data[(data.price>lb) & (data.price<ub)]
```

```python
lb,ub = detect_outliers(data.minimum_nights)
print(lb,ub)
```

```python
data = data[(data.minimum_nights>lb) & (data.minimum_nights<ub)]
```
