# Write a function withdraw(balance, amount) that:

* Returns new balance if amount ≤ balance

* Returns "Insufficient Balance" otherwise

```python
def withdraw(balance,amount):
    new_balance = 0
    if amount<balance:
        new_balance+=balance-amount
        return new_balance
    else:
        return("Insufficient Balance")  
```

## Write a function process_transaction(balance, amount, transaction_type) that supports:

* "deposit" → add amount

* "withdraw" → subtract amount

* "transfer" → subtract amount + 1% transfer fee

### Return the updated balance.
### If amount > balance (for withdraw/transfer), return "Transaction Failed".

```python
def process_transaction(balance,amount,transaction_type):
    if transaction_type == "deposit":
        balance+=amount
    elif transaction_type=="withdraw":
        if amount>balance:
            return "Transection Failed"
        else:
            balance-=amount
    elif transaction_type=="transfer":
        if amount>balance:
            return "Transaction Failed"
        else:
            fee=amount*0.01
            total = amount+fee
            balance-=total
    return balance
```

```python
process_transaction(2000,800,"transfer")
```

## Write a function analyze_results(marks_list) 
### that returns:
    Highest marks

    Lowest marks
    
    Average
    
    Pass rate percentage
    Passing marks = 50
    
### Return the data in a dictionary.

```python
def analyze_results(*marks_list):
    Highest_Marks = max(marks_list) 
    lowest_Marks = min(marks_list)
    average = sum(marks_list)/len(marks_list)
    passed = [m for m in marks_list if m>=50]
    pass_rate = len(passed)/len(marks_list)*100

    result = {
        "Highest" : Highest_Marks,
        "Lowest" : lowest_Marks,
        "Average" : average,
        "Passed" : passed,
        "Passing_Marks" : pass_rate
    }
    return result
```

## Airline Seat Booking With Constraints

### Write a function book_seat(seats, seat_requested):

### Constraints:

    If seat_requested not in seats → return "Invalid Seat"
    
    If seat is "booked" → return "Seat Already Booked"
    
    Otherwise book the seat and return updated seats dictionary

### Example seats dictionary:
{
  "1A": "free",
  "1B": "booked",
  "2A": "free"
}

```python
def book_seat(seats, seat_requested):
    if seat_requested not in seats:
        return "Invalid Seat"
    if seats[seat_requested] == "booked":
        return "Seat Already Booked"
    seats[seat_requested] = "booked"
    return seats
```

## Create a Python program that:

### Defines a function called generate_salary(employee)

####    The employee parameter will be a dictionary containing:

        "name" → employee name
        
        "basic" → basic salary
        
        "overtime_hours" → extra hours worked
        
        "performance" → string ("excellent", "good", "average", "poor")

### The function must:

#####    ✔ 1. Calculate Gross Salary
    
    Gross = basic salary
    
    overtime pay
    
    performance bonus
    
#####    ✔ 2. Overtime Pay Rules
    
    1 overtime hour = 300 PKR
    
####    ✔ 3. Performance Bonus Rules
    
    excellent → 20% of basic
    
    good → 10%
    
    average → 5%
    
    poor → 0%
    
#####    ✔ 4. Apply Deductions
    
    Income tax = 12% of gross
    
    Medical insurance = 1000 PKR
    
#####    ✔ 5. Return & print a full salary slip with this format:
        Employee: Ali
        Basic Salary: 50000
        Overtime Pay: 6000
        Performance Bonus: 10000
        Gross Salary: 66000
        Income Tax (12%): 7920
        Medical Insurance: 1000
        Net Salary: 57080

### 📝 Example Input:
    employee = {
        "name": "Ali",
        "basic": 50000,
        "overtime_hours": 20,
        "performance": "excellent"
    }

### 📝 Example Output:
    Employee: Ali
    Basic Salary: 50000
    Overtime Pay: 6000
    Performance Bonus: 10000
    Gross Salary: 66000
    Income Tax (12%): 7920
    Medical Insurance: 1000
    Net Salary: 57080

```python
def generate_salary(employee):
    net_salary = 0
    bonus = 0
    name = employee['name']
    basic_salary = employee['basic']
    overtime = employee["overtime_hours"]*300    
    if employee['performance']=="excellent":
        bonus+=basic_salary*0.20
    elif employee['performance']=="good":
        bonus+=basic_salary*0.10
    elif employee['performance']=="average":
        bonus+=basic_salary*0.05
    elif employee['performance']=="poor":
        bonus = 0
    gross_salary = basic_salary+overtime+bonus
    income_tax = gross_salary*0.12
    medical_insurance = 1000
    net_salary+=gross_salary-income_tax-medical_insurance

    print(f"Employee : {name}")
    print(f"Basic Salary : {basic_salary}")
    print(f"Over Time Pay : {overtime}")
    print(f"Performance Bonus : {bonus}")
    print(f"Gross Salary : {gross_salary}")
    print(f"Income Tax : {income_tax}")
    print(f"Medical Insurance : {medical_insurance}")
    return f"Net Salary : {net_salary}"
```

## An online store calculates final prices dynamically:

### Function calculate_discount(price, category) → category-based discounts:

    Electronics > $1000 → 15%
    
    Clothing > $500 → 10%
    
    Groceries → 5%
    
    Function apply_coupon(price, coupon_code) → nested function applies coupon:
    
    SAVE10 → $10 off
    
    SAVE20 → $20 off if price > $200
    
    Function final_price(price, category, coupon) → combines discount + coupon + tax (8%)

### 📥 Input Example:
    final_price(1200, "Electronics", "SAVE10")

### 📝 Example Output:
    Price after discount: 1020
    Price after coupon: 1010
    Final Price with Tax: 1090.8

```python
def calculate_discount(price,category):
    discount = 0
    if category=="Electronics" and price>1000:
        discount+=price*0.15
    elif category=="Clothing" and price>500:
        discount+=price*0.10
    elif category=="Groceries":
        discount+=price*0.05
    else:
        discount = 0
    return discount
def apply_coupon(price,coupon_code):
    if coupon_code=="SAVE10":
        return 10
    elif coupon_code=="SAVE20" and price>200:
        return 20
    else:
        return 0
def final_price(price,category,coupon):
    discount = calculate_discount(price,category)
    price_after_discount = price - discount
    print(f"Price After Discount: {price_after_discount}")
    coupon_discount = apply_coupon(price_after_discount,coupon)
    price_after_coupon = price_after_discount-coupon_discount
    print(f"Price After Coupon: {price_after_coupon}")
    tax = price_after_coupon*0.08
    final = price_after_coupon+tax
    print(f"Final Price With Tax: {final}")
```

```python
final_price(1200, "Electronics", "SAVE10")
```

## A hotel calculates room charges:

### Function room_charges(nights, room_type) → rates:

    Standard = $100/night
    
    Deluxe = $200/night
    
    Suite = $400/night
    
    Function additional_services(services) → Breakfast=$20, Spa=$50, Gym=$10
    
    Function apply_membership_discount(amount, membership_years) → 5% discount if membership > 2 years, 10% if >5 years
    
    Function total_bill(nights, room_type, services, membership_years) → prints detailed bill

### 📥 Input Example:
    total_bill(4, "Deluxe", ["Breakfast", "Spa"], 6)

### 📝 Example Output:
Room Charges: 800
Additional Services: 70
Membership Discount: 87
Total Amount: 783

```python
def room_charges(nights, room_type):
    if room_type == "Standard":
        price = nights * 100
    elif room_type == "Deluxe":
        price = nights * 200
    elif room_type == "Suite":
        price = nights * 400
    else:
        price = 0
    return price
def additional_services(services):
    total = 0
    for service in services:
        if service == "Breakfast":
            total += 20
        elif service == "Spa":
            total += 50
        elif service == "Gym":
            total += 10
    return total
def apply_membership(amount, membership_years):
    discount = 0
    if membership_years > 5:
        discount = amount * 0.10
    elif membership_years > 2:
        discount = amount * 0.05
    return discount
    
def total_bill(nights, room_type, services, membership_years):
    room = room_charges(nights, room_type)
    service = additional_services(services)
    sub_total = room + service
    discount = apply_membership(sub_total, membership_years)
    total = sub_total - discount
    print(f"Room Charges: {room}")
    print(f"Additional Services: {service}")
    print(f"MemberShip Discount: {discount}")
    print(f"Total Amount: {total}")
```

```python
total_bill(4, "Deluxe", ["Breakfast", "Spa"], 6)
```

```python
def average_score(marks):
    avg = sum(marks) / len(marks)
    return avg
def grade(avg):
    def assign_grade():
        if avg >= 90:
            return "A"
        elif avg >= 80:
            return "B"
        elif avg >= 70:
            return "C"
        elif avg >= 60:
            return "D"
        else:
            return "F"
    return assign_grade()
def honor_roll(avg, extracurricular):
    if avg >= 85 and extracurricular >= 3:
        return "Yes"
    else:
        return "No"
def student_summary(student):
    name = student["name"]
    marks = student["marks"]
    extracurricular = student["extracurricular"]
    avg = average_score(marks)
    student_grade = grade(avg)
    honor = honor_roll(avg, extracurricular)
    print(f"Name: {name}")
    print(f"Average Score: {avg}")
    print(f"Grade: {student_grade}")
    print(f"Honor Roll: {honor}")
```

## A store applies layered discounts:

    Function base_discount(price) → 5%
    
    Function category_discount(price, category) → Electronics=10%, Clothing=8%, Grocery=3%
    
    Function coupon_discoun
    
    Function final_price(price, category, coupon) → nested calls calculate total discounts + final price

### 📥 Input Example:
    final_price(500, "Electronics", "SAVE10")

### 📝 Example Output:
    Price after base discount: 475
    Price after category discount: 427.5
    Price after coupon: 417.5
    Final Price: 451.3 (with 8% tax)

```python
def base_discount(price):    
    discount = price * 0.05    
    return price - discount

def category_discount(price, category):

    if category == "Electronics":
        discount = price * 0.10
    elif category == "Clothing":
        discount = price * 0.08
    elif category == "Grocery":
        discount = price * 0.03
    else:
        discount = 0
    return price - discount


def coupon_discount(price, coupon):
    if coupon == "SAVE10":
        return price - 10
    elif coupon == "SAVE20":
        return price - 20
    else:
        return price

def final_price(price, category, coupon):

    price_after_base = base_discount(price)
    print(f"Price after base discount: {price_after_base}")

    price_after_category = category_discount(price_after_base, category)
    print(f"Price after category discount: {price_after_category}")

    price_after_coupon = coupon_discount(price_after_category, coupon)
    print(f"Price after coupon: {price_after_coupon}")

    tax = price_after_coupon * 0.08

    final = price_after_coupon + tax

    print(f"Final Price: {final}")
```

```python
final_price(500, "Electronics", "SAVE10")
```
