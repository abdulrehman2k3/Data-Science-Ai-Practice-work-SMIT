## Two team player sets given. Find common players and unique players

```python
Team_One = {"Hussain","Sarim","Talha","Waseem","Saif"}
Team_Two = {"Ali","Hasan","Saif","Taha","Sarim","Saba","Hiba"}
common_player = Team_One.intersection(Team_Two)
unique_player = Team_One.symmetric_difference(Team_Two)
print(Team_One)
print(Team_Two)
print("Common Player:",common_player)
print("Unique Player:",unique_player)
```
