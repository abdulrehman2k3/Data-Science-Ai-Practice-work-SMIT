## A traffic system calculates fines:

Speed limit = 60 km/h
* If speed ≤ 60 → No fine
* If 61–80 → Fine = 500
* If 81–100 → Fine = 2000
* If >100 → License suspended
#### BUT:
*If emergency vehicle → no fine
*If rainy weather AND speed > 60 → fine increases by 50%

```python
speed_limit = 60
vehicle = input("Type of Vehicle Regular/Emergency:")
speed = int(input("Enter your vehicle speed:"))
weather = input("Enter your city weather:")
fine = 0
if vehicle == "emergency":
    print("No Fine")
else:
    if speed<speed_limit:
        print("No fine")
    elif speed>60 and speed<80 and weather!='rainy':
        fine = 500
    elif speed>60 and speed<80 and weather == 'rainy':
        fine = 500+500*0.50
    elif speed>80 and speed<100 and weather!='rainy':
        fine = 2000
    elif speed>80 and speed<=100 and weather=='rainy':
        fine = 2000+2000*0.50
    else:
        print("Licence Suspended")
print(f"Speed:{speed} Fine:{fine}")
```

## Write a program to decide passenger clearance status.
Take input:
* baggage_weight
* passport_valid (Yes/No)
* visa_status (Valid/Invalid)
* criminal_check (Clear/Flagged)
#### Rules:
* baggage > 30kg → extra charge
* passport invalid → reject
* visa invalid → reject
* criminal flagged → security hold
* otherwise → cleared for boarding

```python
baggage_weight = int(input("Enter the wait of baggage:"))
if baggage_weight>=30:
    passport_valid = input("Your passport is valid (Yes/No):")
    if passport_valid == 'yes':
        visa_status = input("What's your vise status (Valid/Invalid):")
        if visa_status == 'valid':
            criminal_check = input("Criminal Record (Clear/Flagged):")
            if criminal_check == 'clear':
                print("Cleared For Boarding")
            else:
                print("Security Hold")
        else:
            print("Your visa status is invalid")
    else:
        print("Your passport is not valid")
else:
    print("Extra wait")
```

## Write a program to detect suspicious banking transactions.
Take input:
* transaction_amount
* location_match (Yes/No)

* previous_transactions_count
* device_verified (Yes/No)
#### Rules:
* amount > 100,000 → suspicious
* location mismatch + high amount → fraud alert
* night transaction + unverified device → high risk
* many transactions (>10 in 1 hour) → freeze account
* otherwise → normal transaction

```python
transaction_amount = int(input("Enter the transaction amount:"))
location = input("Location mismatch (Yes/No):")
time_of_transaction = input("Time of transaction (Day/Night):")
transactions_count = int(input("Enter the previous transactions count:"))
verified =input("Device are verified ? (Yes/No):")

if transaction_amount >100000:
    print("Suspicious")
elif location == "yes" and transaction_amount>100000:
    print("Fraud Alert")
elif time_of_transaction == "night" and verified=="no":
    print("High Risk")
elif transactions_count>10:
    print("Freeze Account")
else:
    print("Normal Transaction")
```

## Write a program to classify a person’s risk level in a police database.
Take input:
* criminal_record (Yes/No)
* number_of_cases
* travel_history_flagged (Yes/No)
* financial_activity_score
* last_activity_days
#### Rules:
* criminal_record = Yes AND cases > 3 → High Risk
* flagged travel + suspicious financial score → High Risk
* no activity for long time (>365 days) → Monitor
* clean record + normal activity → Low Risk
* mixed signals → Medium Risk

```python
criminal_recod = input("Criminal Record (Yes/No):").lower()
number_of_cases = int(input("Enter the number of cases:"))
travel_history_flagged = input("Travel Histry (Yes/No)").lower()
financial_activity_score = int(input("Enter the financial activity score"))
last_activity_days = int(input("Enter the activity days:"))

if (criminal_recod == 'yes' and number_of_cases>3) or (travel_history_flagged=="yes" and financial_activity_score>0):
    print("High Risk")
elif last_activity_days>365:
    print("Monitor")
elif (not(criminal_recod == 'yes' and number_of_cases>3) or (travel_history_flagged=="yes" and financial_activity_score>0)):
    print("Low Risk")
else:
    print("Medium Risk")
```
