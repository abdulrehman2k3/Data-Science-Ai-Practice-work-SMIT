```python
def build_pipeline(degree=1):
    steps = []
    if degree > 1:
        steps.append(("poly", PolynomialFeatures(degree, include_bias=False)))
    steps += [("scaler", StandardScaler()),
              ("model", LinearRegression())]
    return Pipeline(steps)
```
