## Q2 : Write a Python program to create a class called 'Rectangle' with properties for width and height. Include two methods to calculate rectangle area and perimeter. Create an instance of the 'Rectangle' class and calculate its area and perimeter. 

```python
class Rectangle():
    def __init__(self,width,height):
        self.Width = width
        self.Height = height
    def calculate_area(self):
        return f"Rectangle Area : {self.Width*self.Height}"
    def calculate_perameter(self):
        self.perimeter = 2*(self.Width+self.Height)
        return f"Rectangle Perimeter : {self.perimeter}"
```

## Q9 : Write a Python program that creates a class called Bank with properties for bank names and branches. Include methods to add a branch, remove a branch, and display all branches. Create an instance of the Bank class and perform operations to add and remove branches. 

```python
class Bank():
    def __init__(self,bank_name,branches):
        self.bank_name = bank_name
        self.branches = [branches]
    def add_branch(self,branch_names):
        self.branches.append(branch_names)
        return "Branches Added Successfully"
    def remove_branch(self,branch_name):
        for branch in self.branches:
            if branch_name in branch:
                self.branches.remove(branch_name)
                return "Branch Remove Successfully"
    def display_all_branches(self):
        return f"All Branches of {self.bank_name} bank is {self.branches}"
```

```python
b1 = Bank("Meezan","Gulshan Branch")
print(b1.display_all_branches())
```

```python
b1.add_branch("Hydri Branch")
b1.add_branch("Sadar Branch")
b1.add_branch("Karsaz Branch")
```

```python
class product():
    def __init__(self,product_id,name,price,quantity):
        self.product_id = product_id
        self.name = name
        self.price = price
        self.quantity = quantity
    def total_price(self):
        total_price = self.quantity*self.price
        return f"Total Price : {total_price}"
class PersonalCareProduct(product):
    def __init__(self,product_id,name,price,quantity,warranty_period):
        super().__init__(product_id,name,price,quantity)
        self.warranty_period = warranty_period
    def total_price(self):
        total = self.price*self.quantity
        
        if self.warranty_period==0:
            self.warranty_period=0
            
        elif self.warranty_period==1:
            extra_charges = total*0.05
            final = total+extra_charges
            
        elif self.warranty_period>1 and self.warranty_period<3:
            extra_charges = total*0.10
            final = total+extra_charges
        else:
            extra_charges = total*0.50
            final = total+extra_charges
        return f"Total Price with Warranty charges is {final}"
```

## Q11 : Write a Python program that creates a class called BankAccount with properties for account number, account holder name, and balance. Include methods to deposit, withdraw, and transfer money between accounts. Create multiple instances of the BankAccount class and perform operations such as depositing, withdrawing, and transferring money. 

```python
class BankAccount():
    def __init__(self,account_holder_name,initial_balance):
        self.account_no = random.randint(22222222,99999999)
        self.account_name = account_holder_name
        self.__balance = initial_balance
    def deposit(self,amount):
        if amount>0:
            self.__balance+=amount
            return "Amount Deposit Successfully"
        else:
            print("Negative amount not allowed!")
    def withdraw(self,amount):
        if amount>0:
            if amount<self.__balance:
                self.__balance-=amount
                return "Withdraw money successfully"
            else:
                print("Insufficient Balance!")
        else:
            return "Sorry, Negative Amount is not able to deposit"
    def transfer_money(self,other,amount):
        if amount>0:
            if amount<self.__balance:
                self.__balance-=amount
                other.__balance+=amount
                return "Money Transfer Successfully"
            else:
                print("Insufficient Balance!")
        else:
            print("Negative ammount not allowed!")
    def show_balance(self):
        return f"Mr {self.account_name} your current balance is {self.__balance}"
```

## Q12 : Write a Python program that creates a class called University with properties for university name and departments. Include methods to add a department, remove a department, and display all departments. Create an instance of the University class and add and remove departments.

```python
class University():
    def __init__(self,name,departments):
        self.name = name
        self.department = [departments]
    def add_departments(self,new_department):
        self.department.append(new_department)
        return "New department added successfully"
    def remove_department(self,department):
        self.department.remove(department)
        return "Remove Department Successfully"
    def Show_departments(self):
        return f"All Departments {self.department}"
        
```
