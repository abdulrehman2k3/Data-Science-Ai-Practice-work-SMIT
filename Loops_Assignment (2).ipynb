## Create a menu-driven banking system using loops:
* Create Account
* Deposit
* Withdraw
* Check Balance
* Exit

```python
name = ""
balance = 0
account_created = False
for i in range(100):
    print("""
=== Banking System ===
1. Create Account
2. Deposit
3. Withdraw
4. Check Balance
5. Exit
    """)
    choice = int(input("Enter your choice: "))
    if choice == 1:
        name = input("Enter your name: ")
        balance = 0
        account_created = True
        print(f"Account created for {name}")

    elif choice == 2:
        if not account_created:
            print("Please create account first")
            continue
        amount = int(input("Enter deposit amount: "))
        if amount <= 0:
            print("Invalid amount")
            continue
        balance += amount
        print(f"Rs.{amount} deposited")
    elif choice == 3:
        if not account_created:
            print("Please create account first!")
            continue
        amount = int(input("Enter withdraw amount: "))
        if amount <= 0:
            print("Invalid amount!")
        elif amount > balance:
            print("Insufficient balance!")
        else:
            balance -= amount
            print(f"Rs.{amount} withdrawn!")
    elif choice == 4:
        if not account_created:
            print("Please create account first!")
            continue
        print(f"Account Holder: {name}")
        print(f"Current Balance: Rs.{balance}")
    elif choice == 5:
        print("Thank you for using our Banking System!")
        break
    else:
        print("Invalid choice! Try again.")
```

## School Marks Report
### Goal:

    Take input for multiple students’ marks in 3 subjects and calculate their total and average.

### Steps:

    Outer loop → for each student (3 students).
    
    Inner loop → take marks for 3 subjects.
    
    Store each student’s total in a list.
    
    Print all totals and class average.

### Example:
    Enter marks for Student 1:
    Subject 1: 80
    Subject 2: 75
    Subject 3: 90
    Enter marks for Student 2:
    Subject 1: 60
    Subject 2: 70
    Subject 3: 65
    → Output: [245, 195]

```python
all_totals = []
for student in range(1, 4):
    total = 0
    print(f"Student {student}")
    for subject in range(1, 4): 
        marks = int(input(f"  Subject {subject} marks: "))
        total += marks
    all_totals.append(total)
    average = total / 3
    print(f"  Total: {total} | Average: {average:.2f}")

class_average = sum(all_totals) / len(all_totals)
print(f"All Totals: {all_totals}")
print(f"Class Average: {class_average:.2f}")
```

## Shop Sales Tracker
### Goal:

    Track daily sales of 3 products over 5 days and calculate the total revenue.

### Steps:

    Outer loop → iterate over 5 days.
    
    Inner loop → take sales amount for 3 products each day.
    
    Store daily total in a list.
    
    If any product sale = 0 → continue (skip that product).
    
    Print all daily totals and grand total.

### Example:
    Day 1:
    Product 1: 200
    Product 2: 150
    Product 3: 0 → skipped
    Day 2:
    Product 1: 300
    Product 2: 250
    Product 3: 400
    → Daily Totals = [350, 950], Grand Total = 1300

```python
daily_totals = []
for day in range(1, 6):
    daily_total = 0
    print(f"Day {day}")
    for product in range(1, 4): 
        sale = int(input(f"  Product {product} sales: "))
        if sale == 0:
            print("  Skipped!")
            continue
        daily_total += sale
    daily_totals.append(daily_total)
    print(f"  Day {day} Total: {daily_total}")

grand_total = sum(daily_totals)
print(f"Daily Totals: {daily_totals}")
print(f"Grand Total: {grand_total}")
```

## Invoice Tax Calculator
### Goal:

Take multiple items in multiple invoices and calculate tax per invoice.

Outer loop → for 3 invoices.

Calculate total + 10% tax.

```python
subjects = ['Math', 'English', 'Science', 'Urdu']
for i in range(1, 6): 
    total = 0
    failed = False
    print(f"Student {i} ")
    for subject in subjects: 
        marks = int(input(f"  Enter {subject} marks: "))
        if marks < 33:
            print(f"  Failed in {subject}!")
            failed = True
        total += marks
    average = total / 4
    if average >= 80:
        grade = 'A'
    elif average >= 60:
        grade = 'B'
    elif average >= 45:
        grade = 'C'
    elif average >= 33:
        grade = 'D'
    else:
        grade = 'F'
    print(f"  Total: {total} | Average: {average:.2f} | Grade: {grade}")
    if failed:
        print(f"  Status: Failed in one or more subjects!")
    else:
        print(f"  Status: Pass")
```

## Holiday Package Evaluator
### Goal:

Compare multiple travel packages across features like hotel, transport, food.

### Steps:

Outer loop → 3 packages.

Inner loop → 3 features.

Input rating out of 5.

Skip if rating = 0.

Print average rating per package.

### Example:
Package 1: [4, 5, 5] → Avg = 4.67  
Package 2: [3, 4, 0] → Avg = 3.5

```python
features = ['Hotel', 'Transport', 'Food']
for i in range(1, 4):  
    total = 0
    count = 0
    print(f"Package {i}:")
    for feature in features:   
        rating = float(input(f"  Enter {feature} rating (0-5): "))       
        if rating == 0:
            continue   
        total += rating
        count += 1
    if count > 0:
        average = total / count
        print(f"Average rating of Package {i}: {average:.2f}")
    else:
        print("No ratings given!")
```
